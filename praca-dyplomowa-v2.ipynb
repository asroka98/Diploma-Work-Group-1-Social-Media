```python
dfQ = df.copy()

dfQ['User_ID'] = ['U' + str(i+1).zfill(3) for i in range(len(dfQ))]
dfQ.loc[dfQ['Age'] > 110, 'Age'] = pd.NA
dfQ['Gender'] = df['Gender'].replace('ERR', pd.NA)
dfQ.loc[dfQ['Daily_Screen_Time(hrs)'] > 11, 'Daily_Screen_Time(hrs)'] = pd.NA
dfQ.loc[dfQ['Sleep_Quality(1-10)'] > 10, 'Sleep_Quality(1-10)'] = pd.NA
dfQ.loc[dfQ['Stress_Level(1-10)'] > 10, 'Stress_Level(1-10)'] = pd.NA
dfQ.loc[dfQ['Days_Without_Social_Media'] > 7, 'Days_Without_Social_Media'] = pd.NA
#Exercise_Frequency(week)  dfQ.loc[dfQ['Exercise_Frequency(week)'] > 7, 'Exercise_Frequency(week)'] = pd.NA
dfQ = dfQ[dfQ['Social_Media_Platform'] != 'ERR']
dfQ.loc[dfQ['Happiness_Index(1-10)'] > 10, 'Happiness_Index(1-10)'] = pd.NA

dfQ
```

```python
dfQ.info()
```

```python
dfQ['Age'] = np.ceil(dfQ['Age']).astype('Int64')
dfQ['Sleep_Quality(1-10)']=dfQ['Sleep_Quality(1-10)'].astype('Int64')
dfQ['Stress_Level(1-10)']=dfQ['Stress_Level(1-10)'].astype('Int64')
dfQ['Days_Without_Social_Media']=dfQ['Days_Without_Social_Media'].astype('Int64')
dfQ['Exercise_Frequency(week)'] = np.ceil(dfQ['Exercise_Frequency(week)']).astype('Int64')
dfQ['Happiness_Index(1-10)']=dfQ['Happiness_Index(1-10)'].astype('Int64')
dfQ
```
